## 1. Trend analysis

```python
def clean_percentage_column(series):
    return (
        series.astype(str)
        .str.replace(">", "", regex=False)
        .str.replace("<", "", regex=False)
        .str.strip()
        .replace("nan", pd.NA)
        .astype(int, errors="ignore")
    )

water_data_clean["at_least_basic"] = clean_percentage_column(
    water_data_clean["at_least_basic"]
)
```

```python
import matplotlib.pyplot as plt

selected_countries = [
    "Morocco",
    "France",
    "United States of America",
    "China"
]

filtered_df = water_data_clean[
    water_data_clean["country"].isin(selected_countries)
]

plt.figure(figsize=(20,15))

for country in selected_countries:
    subset = filtered_df[filtered_df["country"] == country]
    plt.plot(subset["year"], subset["at_least_basic"], label=country)
    plt.legend()
    plt.title("Access to Basic Drinking Water")
plt.show()
```

### Interpretation of the “Access to Basic Drinking Water” plot
- **France** is essentially at universal access across the whole period. The dataset shows >99% for every year from 2000 to 2024, so its line is flat at the top of the chart.

- **United States of America** also sits near the top. Early years in the raw data are missing, but from 2005 onward it is 98.9% or higher and becomes >99% for much of the series.

- **China** shows a strong upward trend:
    - around 80.6% in 2000
    - rising steadily each year
    - reaching about 96.1% by 2024

- **Morocco** shows the largest relative improvement:
    - about 62.1% in 2000
    - a steady increase over time
    - reaching about 92.2% in 2024

## 2. Urbanization vs Access Correlation

```python
def clean_percentage_column(series):
    return (
        series.astype(str)
        .str.replace(">", "", regex=False)
        .str.replace("<", "", regex=False)
        .str.strip()
        .replace("nan", pd.NA)
        .astype(int, errors="ignore")
    )

water_data_clean["%_urban"] = clean_percentage_column(
    water_data_clean["%_urban"]
)
```

### Interpretation of the “Urbanization vs Water Access” plot
- **Overall relationship** : There is a moderate positive association between urbanization and basic water access.

- **Correlation**: In the dataset, %_urban and at_least_basic have a correlation of about `0.56`, which means higher urban share generally goes with higher water access.

- **Distribution shape** :
    - The average country in the dataset has about `59%` urban population and `87%` basic water access.
    - Many data points are clustered near very high access values `(95–99%)`, even across a wide range of urbanization.

- **Key pattern** :
    - Low-urbanization cases still can show very high access, especially for small territories like Tokelau and Wallis and Futuna, which appear at `0%` urbanization but `99%` water access.
    - High-urbanization countries also frequently sit at `99%` access, so the right edge of the plot is dense near the top.

## 4. Distribution Analysis

```python
import matplotlib.pyplot as plt

latest_year01 = water_data_clean["year"].max()

top10_water = (
    water_data_clean[water_data_clean["year"] == latest_year01]
    .sort_values("at_least_basic", ascending=True)
    .head(10)
)

plt.figure(figsize=(12,6))
plt.bar(top10_water["country"], top10_water["at_least_basic"])

plt.title(f"Top 10 Countries by Basic Drinking Water Access ({latest_year01})")
plt.xlabel("Country")
plt.ylabel("% Population with Access")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Interpretation of the "Top 10 Countries by Basic Drinking Water Access (2024)" plot

- **The chart shows the 10 countries with the lowest basic drinking water access in 2024**, not the highest. This is the bottom tier globally.

- **Access ranges from approximately 50% to 80%** among these countries, meaning a significant portion of their populations still lack access to basic drinking water.

- **Key patterns**:
  - The lowest-performing country sits around 50%, indicating that half its population lacks basic water access.
  - Most countries in the bottom 10 are between 60% and 80%, showing they are still developing water infrastructure.
  - There is a notable gap between these countries and the global high performers (France, USA, China, which are >96%).

- **Geographic/developmental context**: These bottom-10 countries typically face infrastructure challenges, limited investment in water systems, or geographical constraints that make water access difficult.

- **Key takeaway**: While developed nations have achieved near-universal basic water access, these 10 countries represent the global challenge—roughly half to one-third of their populations still lack reliable access to basic drinking water, indicating a significant development need.

## 7. Multi-Dataset Comparison

```python
comparison_df = (
    water_data_clean[["country", "year", "at_least_basic"]]
    .merge(
        sanitation_data_clean[["country", "year", "at_least_basic"]],
        on=["country", "year"],
        suffixes=("_water", "_sanitation")
    )
    .merge(
        hygiene_data_clean[["country", "year", "basic"]],
        on=["country", "year"]
    )
    .rename(columns={"basic": "hygiene"})
)
```

```python
comparison_df
```

### Interpretation of the "WASH Dataset Sample"

- **The dataset spans multiple decades and countries**, with Afghanistan showing a clear upward trend in water and sanitation access from 2005 onward, while hygiene values remain relatively flat — suggesting infrastructure investment outpaced hygiene improvements.

- **NaN values in the hygiene column for Zimbabwe** highlight a common data completeness issue, where hygiene data is often missing for earlier years and lower-income countries, limiting analysis for that dimension.

- **Key takeaway** : The dataset captures long-term WASH trends but suffers from hygiene data gaps, particularly for African countries in earlier periods — a limitation to keep in mind when interpreting hygiene results.

```python
import matplotlib.pyplot as plt
import numpy as np

latest_year = comparison_df["year"].max()

selected = comparison_df[
    (comparison_df["country"].isin([
        "Morocco",
        "France",
        "China",
        "United States of America"
    ])) &
    (comparison_df["year"] == latest_year)
]

x = np.arange(len(selected))
width = 0.25

plt.figure(figsize=(12,6))

plt.bar(
    x - width,
    selected["at_least_basic_water"],
    width,
    label="Water"
)

plt.bar(
    x,
    selected["at_least_basic_sanitation"],
    width,
    label="Sanitation"
)

plt.bar(
    x + width,
    selected["hygiene"],
    width,
    label="Hygiene"
)

plt.xticks(x, selected["country"])
plt.ylabel("Access Rate (%)")
plt.title(f"WASH Comparison by Country ({latest_year})")
plt.legend()

plt.tight_layout()
plt.show()
```

### Interpretation of the "WASH Comparison by Country (2024)"

- **All three countries — China, France, and the USA — achieve near-universal water and sanitation access (~99%)**, confirming that high-income and upper-middle-income nations have essentially solved basic WASH coverage.

- **China is the only country displaying a visible hygiene bar (~97%)**, while France and the USA appear to have no hygiene data reported, likely due to missing values rather than actual zero access.

- **Key takeaway** : This chart confirms that developed nations have converged toward full WASH coverage, making within-group differences negligible. The missing hygiene data for France and the USA is a data reporting gap, not a reflection of real-world conditions.
